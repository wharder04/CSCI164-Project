```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset 1
df1 = pd.read_csv("Gaming_Academic_Performance.csv")

print("Shape:", df1.shape)
print("\nColumns:")
print(df1.columns.tolist())

print("\nData types:")
print(df1.dtypes)

print("\nMissing values:")
print(df1.isnull().sum())

display(df1.head())
```

```python
Dataset Structure:
- **Rows:** 8000
- **Columns:** 14
- **Target Variable:** `grades`
- **Task Type:** Regression

Features:
- `gaming_hours`
- `study_hours`
- `sleep_hours`
- `attendance`
- `social_activity`
- `device_usage`
- `reaction_time_ms`
- `addiction_score`
- `stress_level`
- `gender`
- `gaming_genre`
```

```python
#Drop Id
df1 = df1.drop(columns=["student_id"])

#Features
X1 = df1.drop(columns=["grades"])
y1 = df1["grades"]

#Types
numeric_features_1 = X1.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features_1 = X1.select_dtypes(include=["object"]).columns.tolist()

print("Numeric features:", numeric_features_1)
print("Categorical features:", categorical_features_1)
```

```python
#Processing Plan
Steps:
1. Remove indentifier column (student id)
2. Missing values (no missing values present)
3. Encode catergorical variables (gender/gaming genre converted with hot-one)
4. Normalize numerical variables
```

```python
Dataset 1: AI_Impact_Student_Life_2026.csv
The dataset was selected because it focus on the same idea with students, but about ai usage. 
It contains both numerical and categorical features, no missing values again, and strong
targets and GPA change with ai usage.
```

```python
# Dataset 2
df2 = pd.read_csv("AI_Impact_Student_Life_2026.csv")

print("Shape:", df2.shape)
print("\nColumns:")
print(df2.columns.tolist())

print("\nData types:")
print(df2.dtypes)

print("\nMissing values:")
print(df2.isnull().sum())

display(df2.head())
```

```python
# Create GPA change feature
df2["GPA_Change"] = df2["GPA_Post_AI"] - df2["GPA_Baseline"]

display(df2[["GPA_Baseline", "GPA_Post_AI", "GPA_Change"]].head())
```

```python
## Dataset 2 Structure
- **Rows:** 1500
- **Columns:** 11 original columns (plus 1 derived feature: `GPA_Change`)
- **Target Variable:** `GPA_Change` (preferred) or `GPA_Post_AI`
- **Task Type:** Regression

###Features:
- `Age`
- `Major`
- `Primary_AI_Tool`
- `Task_Frequency_Daily`
- `Main_Usage_Case`
- `GPA_Baseline`
- `Time_Saved_Hours_Weekly`
- `AI_Ethics_Concern`
- `Career_Confidence_Score`
```

```python
# DropID
df2 = df2.drop(columns=["Student_ID"])

#GPA_Change as target
X2 = df2.drop(columns=["GPA_Change"])
y2 = df2["GPA_Change"]

#Types
numeric_features_2 = X2.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features_2 = X2.select_dtypes(include=["object"]).columns.tolist()

print("Numeric features:", numeric_features_2)
print("Categorical features:", categorical_features_2)
```

```python
Processing Plan
Steps:
1. Remove indentifier column (student id)
2. Feature working (GPA_change - base and ai usage)
3. Missing values (no missing values)
4. Encode Categorical values (major, primary ai tool, main usage case)
5. Normalize numerical variables
```

```python
Dataset 1 Prior work:
Mahmud et al. (2023), Online gaming and its effect on academic performance, sleep, and physical activity in university students.
https://pmc.ncbi.nlm.nih.gov/articles/PMC10723781/
This study is relevant to the project because the dataset used is very similar to the one I am using with similar variables.
The study reported that students who played games for long periods, especially 30+ hours per week, tended to have lower GPA, 
less physical activity, and less study time. This supports the use of the `Gaming_Academic_Performance.csv` dataset 
for predicting grades from gaming and lifestyle factors.
```
